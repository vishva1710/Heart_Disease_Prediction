***4.Data Preprocessing***

```python
#segrigating x and y
x=heart_dis.drop(columns=[TARGET_COL])
y=heart_dis[TARGET_COL]

#create a signature for each feature row-to prevent duplicates flow to test set
row_signature=pd.util.hash_pandas_object(x,index=False)
```

```python
#split data into train and test sets using GroupShuffleSplit
grps=GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=RANDOM_STATE
)
train_indx,test_indx=next(
    grps.split(x,y,groups=row_signature)
)
x_train,x_test=x.iloc[train_indx],x.iloc[test_indx]
y_train,y_test=y.iloc[train_indx],y.iloc[test_indx]

print("Train shape:",x_train.shape)
print("Test shape:",x_test.shape)
```

***Groupshufflesplit:***
My dataset contains many repeated feature combinations. If I use a normal random split, identical observations can appear in both training and validation data, which can cause data leakage and overly optimistic performance. So I used group-based splitting to keep the same group entirely within either training or validation data.

```python
print("dataset_size:",x.shape)
print("Train_data_size:",x_train.shape)
print("Test_dataset_size:",x_test.shape)
```

***5.Baseline Model-Logistic Regression***

```python
base_pipeline=Pipeline(
    steps=[
        ("scaler",StandardScaler()),
        ("model",LogisticRegression())
    ]
)
```

```python
#training the base pipeline
base_pipeline.fit(x_train,y_train)
```

```python
#resuable evaluate function
def evaluate_classifier(model,x_train,y_train,x_test,y_test,model_name):
    #prediction
    y_train_predict=model.predict(x_train)
    y_test_predict=model.predict(x_test)

    #accuracy
    train_accuracy=accuracy_score(y_train,y_train_predict)*100
    test_accuracy=accuracy_score(y_test,y_test_predict)*100

    #results
    print(f"{model_name}-Train Accuracy{train_accuracy:.2f}%")
    print(f"{model_name}-Test Accuracy{test_accuracy:.2f}%\n")

    print("-"*40)

    print("train classificatoion report")
    print(classification_report(y_train,y_train_predict))
    print("-"*40)

    print("test classification report")
    print(classification_report(y_test,y_test_predict))
```

```python
evaluate_classifier(
    base_pipeline,
    x_train,
    y_train,
    x_test,
    y_test,
    model_name="Logistic Regression -Base Model"
)
```

***6.Model Selection***

```python
#models to evaluate
models={
    "Logistic Regression":LogisticRegression(random_state=RANDOM_STATE),
    "Support Vector Machine":SVC(random_state=RANDOM_STATE),
    "Random Forest":RandomForestClassifier(random_state=RANDOM_STATE),
    "XGBoost":XGBClassifier()
}
```

```python
groups=pd.util.hash_pandas_object(x_train,index=False)
k=5
cv=StratifiedGroupKFold(n_splits=k,shuffle=True,random_state=RANDOM_STATE)
```

***StratifiedgroupKfold:***
I used StratifiedGroupKFold because I wanted to satisfy two requirements simultaneously. First, observations belonging to the same group should not be split across folds. Second, the target class distribution should remain approximately balanced across the folds. This gives a more reliable cross-validation estimate.

***Why not normal StratifiedKFold?:***
StratifiedKFold maintains the target distribution, but it does not consider groups. Therefore, rows from the same group could appear in both training and validation folds. StratifiedGroupKFold handles both group separation and class stratification.

```python
for name ,model in models.items():
    recall_scores=[]
    f1_scores=[]
    for tr_idx,te_idx in cv.split(x_train,y_train,groups=groups):
        x_tr,x_te=x_train.iloc[tr_idx],x_train.iloc[te_idx]
        y_tr,y_te=y_train.iloc[tr_idx],y_train.iloc[te_idx]

        pipeline=Pipeline(
            steps=[
                ("scaler",StandardScaler()),
                ("model",model)
            ]
        )
        pipeline.fit(x_tr,y_tr)
        pred=pipeline.predict(x_te)

        recall_scores.append(round(recall_score(y_te,pred),2))
        f1_scores.append(round(f1_score(y_te,pred),2))
    print("Model Name:",name)
    print("Recall Values",recall_scores)
    print("F1 scores:",f1_scores)
    print("\nCV Recall mean:",round(float(np.mean(recall_scores)),4))
    print("CV F1 scores mean:",round(float(np.mean(f1_scores)),4))
    print("-"*50)

#cross validation is a more realiable validation technique than a simple train test split
```

***RandomForestClassifier provides high performance***

**Hyperparameter Tuning -RFC**

```python
rfc_pipe=Pipeline(
    steps=[
        ("scaler",StandardScaler()),
        ("model",RandomForestClassifier(
            random_state=RANDOM_STATE,
            n_jobs=-1
        ))

    ]
)
```

```python
#parameter distribution
param_dist={
    "model__n_estimators":randint(400,1200),
    "model__max_depth":[3,4,5,6],
    "model__min_samples_split":randint(10,40),
    "model__min_samples_leaf":randint(4,20),
    "model__max_features":[0.25,0.35,0.5,"sqrt"],
    "model__bootstrap":[True],
    "model__max_samples":uniform(0.6,0.35),
    "model__ccp_alpha":uniform(0.0,0.02),
}
```

```python
# randomized search
random_search = RandomizedSearchCV(
    estimator=rfc_pipe,
    param_distributions=param_dist,
    n_iter=60,
    scoring="f1_macro",
    cv=cv,
    random_state=RANDOM_STATE,
    verbose=2,
    n_jobs=-1
)
```

```python
#performing randomizedsearchcv
random_search.fit(x_train,y_train,groups=groups)
```

```python
print("Hyperparamter Tuning Result:")
print("best params:",random_search.best_params_)
print("Best score:",random_search.best_score_)
```

***7.Retrain with best params***

```python
rfc_best=Pipeline(
    steps=[
        ("scale",StandardScaler()),
        ("model",RandomForestClassifier(
            random_state=RANDOM_STATE,
            n_jobs=-1,
            n_estimators=random_search.best_params_["model__n_estimators"],
            max_depth=random_search.best_params_["model__max_depth"],
            min_samples_split=random_search.best_params_["model__min_samples_split"],
            min_samples_leaf=random_search.best_params_["model__min_samples_leaf"],
            max_features=random_search.best_params_["model__max_features"],
            bootstrap=random_search.best_params_["model__bootstrap"]

        ))
    ]
)
```

```python
#train the best model
rfc_best.fit(x_train,y_train)
```

***8.Model Evaluation***

```python
evaluate_classifier(
    rfc_best,
    x_train,
    y_train,
    x_test,
    y_test,
    model_name="Random Forest - Best Model"
)
```

***9.Build a Predictive system***

```python
def predict_heart_disease(input_features):

    #create dataframe with correct column order
    input_df=pd.DataFrame(
        [input_features],
        columns=x_train.columns
    )

    #class prediction
    predict=rfc_best.predict(input_df)[0]

    #probablity prediction (positive class)
    prob=rfc_best.predict_proba(input_df)[0][1]

    #output
    print(f"Predicted Class:{predict}")
    print(f"Heart disease risk probability:{prob:.2%}")

    if predict==1:
        print("Diagnosis:High risk of heart disease😢")
    else:
        print("Diagnosis:low risk of heart disese")
    return predict,prob
```
